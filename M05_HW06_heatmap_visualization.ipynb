## Problem 5
Use the Pokemon dataset to create a clustermap with color. First, filter the dataframe to only keep data with `Type 1` equals to following values: `Water`, `Normal`, `Grass`, `Bug` and `Psychic`. Annotate the dendrogram using different colors for these five different `Type 1` values. Use `Name` as the index.

pokemon_data is in https://raw.githubusercontent.com/csbfx/advpy122-data/master/Pokemon.csv

```python
# Problem 5
# Your code here . . .
pokemon_data = pd.read_csv("https://raw.githubusercontent.com/csbfx/advpy122-data/master/Pokemon.csv")
pokemon_data
```

```python
## filter dataframe to only keep data with type 1 types water, normal, grass, bug, and physics
allowed_types = ['Water', 'Normal', 'Grass', 'Bug', 'Psychic']
filtered_pokemon_data = pokemon_data[pokemon_data['Type 1'].isin(allowed_types)]
filtered_pokemon_data
```

```python
## set name of pokemon as index
filtered_pokemon_data.set_index('Name', inplace=True, drop=True)
filtered_pokemon_data
```

```python
# only include numerical columns (int/float)
numerical_cols = filtered_pokemon_data.select_dtypes(include=['number'])
columns_to_drop = ['#', 'Generation', 'Legendary', 'Total']
data_for_clustermap = numerical_cols.drop(columns=columns_to_drop, errors='ignore')
data_for_clustermap
```

```python
# color palette for the 'Type 1' column
type_colors = {
    'Water': 'blue',
    'Normal': 'grey',
    'Grass': 'green',
    'Bug': 'lime',
    'Psychic': 'purple'
}

# make list of colors for the rows based on 'Type 1'
row_colors = filtered_pokemon_data['Type 1'].map(type_colors)

# clustermap
g = sns.clustermap(data_for_clustermap, cmap='viridis', figsize=(10, 10), row_colors=row_colors)
plt.suptitle('Clustermap of Pokemon Stats by Type 1', y=1.02)

# custom legend for row colors
import matplotlib.patches as mpatches
patches = [mpatches.Patch(color=color, label=label) for label, color in type_colors.items()]
plt.legend(handles=patches, title='Type 1', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.show()
```
